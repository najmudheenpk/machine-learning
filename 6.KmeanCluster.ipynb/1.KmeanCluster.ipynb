### Income Clustering

```python
import pandas as pd 
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('income.csv')
df.head()
```

```python
import matplotlib.pyplot as plt 
plt.scatter (df.Age,df['Income($)'])
plt.xlabel('Age')
plt.ylabel('Income')
```

```python
from sklearn.preprocessing import MinMaxScaler 
scaler = MinMaxScaler()

df['Age'] = scaler.fit_transform(df[['Age']])
df['Income($)'] = scaler.fit_transform(df[['Income($)']])
df.head()
```

```python
# finding number of clusters using elbow method 

sse = []
k_rng = range(1,10)
for k in k_rng:
    km = KMeans(n_clusters=k)
    km.fit(df[['Age','Income($)']])
    sse.append(km.inertia_)
```

```python
plt.xlabel('K')
plt.ylabel('Sum of squared error')
plt.plot(k_rng,sse)
```

```python
model = KMeans(n_clusters=3)
y_predicted = model.fit_predict(df[['Age','Income($)']])
y_predicted
```

```python
df['cluster'] = y_predicted
df.head()
```

```python
model.cluster_centers_
```

```python

df1 = df[df.cluster==0]
df2 = df[df.cluster==1]
df3 = df[df.cluster==2]
plt.scatter(df1.Age,df1['Income($)'],color='green')
plt.scatter(df2.Age,df2['Income($)'],color='red')
plt.scatter(df3.Age,df3['Income($)'],color='black')
plt.scatter(model.cluster_centers_[:,0],model.cluster_centers_[:,1],color='purple',marker='*',label='centroid')
plt.xlabel('Age')
plt.ylabel('Income ($)')
plt.legend()
```

### Iris Clustering 

```python
import pandas as pd 
from sklearn.datasets import load_iris 
iris = load_iris()
dir(iris)
```

```python
data = pd.DataFrame(iris.data,columns=iris.feature_names)
data.head()
```

```python
data['flower'] = iris.target 
data.head()
```

```python
from sklearn.cluster import KMeans
km_range= range(1,10)
errors = []
for i in km_range:
    cluster = KMeans(n_clusters=i)
    cluster.fit(data[['petal length (cm)', 'petal width (cm)']])
    errors.append(cluster.inertia_)
```

```python
plt.plot(km_range,errors)
plt.xlabel('Number of K')
plt.ylabel('sum of squared error')
```

```python
cluster_model = KMeans(n_clusters=3)
predicted_cluster = cluster_model.fit_predict(data[['petal length (cm)', 'petal width (cm)']])
data['Clusters'] = predicted_cluster
```
