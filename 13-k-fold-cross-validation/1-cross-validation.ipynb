```python
from sklearn.linear_model  import LogisticRegression
from sklearn.svm import SVC 
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
digits = load_digits()
```

```python
scaler = MinMaxScaler()
X = scaler.fit_transform(digits.data)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,digits.target,test_size=0.3)
```

Logistic Regression

```python
lr = LogisticRegression(max_iter=100)
lr.fit(X_train,y_train)
lr.score(X_test,y_test)
```

SVM

```python
svm = SVC()
svm.fit(X_train,y_train)
svm.score(X_test,y_test)
```

```python
rf = RandomForestClassifier(n_estimators=100)
rf.fit(X_train,y_train)
rf.score(X_test,y_test)
```

```python
def get_score(model,X_train,X_test,y_train,y_test):
    model.fit(X_train,y_train)
    return model.score(X_test,y_test)
```

```python
from sklearn.model_selection import KFold
folds = KFold(n_splits=3)

scores_logistic = []
scores_svm = []
scores_random = []
```

```python
for train_index , test_index in folds.split(X, digits.target):
    X_train , X_test , y_train , y_test = X[train_index] , X[test_index], \
        digits.target[train_index] , digits.target[test_index]
        
    scores_logistic.append(get_score(LogisticRegression(),X_train , X_test , y_train , y_test))
    scores_svm.append(get_score(SVC(),X_train , X_test , y_train , y_test))
    scores_random.append(get_score(RandomForestClassifier(n_estimators=100),X_train , X_test , y_train , y_test))
```

```python
scores_logistic
```

```python
np.mean(scores_logistic)
```

```python
scores_svm
```

```python
np.mean(scores_svm)
```

```python
scores_random
```

```python
np.mean(scores_random)
```

### cross_val_score function

```python
from sklearn.model_selection import cross_val_score
```

```python
#logistic regression

cross_val_score(LogisticRegression() , X, digits.target , cv=5)
```

```python
#svm

cross_val_score(SVC(), X, digits.target , cv=5 )
```

```python
# radom forest

cross_val_score(RandomForestClassifier(n_estimators=100), X , digits.target , cv=5)
```

### Parameter Tuning using K-fold 

```python
score1 = cross_val_score(RandomForestClassifier(n_estimators=5), X , digits.target , cv=5)
np.average(score1)
```

```python
score2 = cross_val_score(RandomForestClassifier(n_estimators=20), X , digits.target , cv=5)
np.average(score2)
```

```python
score3 = cross_val_score(RandomForestClassifier(n_estimators=50), X , digits.target , cv=5)
np.average(score3)
```

```python
score4 = cross_val_score(RandomForestClassifier(n_estimators=100), X , digits.target , cv=5)
np.average(score4)
```
