```python
from sklearn.datasets import load_iris 
from sklearn.model_selection import cross_val_score 
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier 
from sklearn.svm import SVC 
import numpy as np 
```

```python
iris = load_iris()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(iris.data)
```

Logistic Regression

```python
lr_scores = cross_val_score(LogisticRegression(),scaled_data , iris.target, cv=5)
lr_scores
```

```python
np.average(lr_scores)
```

Support Vector Machine

```python
svm_scores = cross_val_score(SVC(),scaled_data, iris.target , cv=5)
svm_scores
```

```python
np.average(svm_scores)
```

Decision Tree

```python
tree_score = cross_val_score(DecisionTreeClassifier(), scaled_data , iris.target , cv=5)
tree_score
```

```python
np.average(tree_score)
```

Random Forest

```python
forest_score = cross_val_score (RandomForestClassifier() , scaled_data , iris.target , cv=5)
forest_score
```

```python
np.average(forest_score)
```
