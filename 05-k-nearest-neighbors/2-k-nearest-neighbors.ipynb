### Digit Classification 

```python
import pandas as pd 
from sklearn.datasets import load_digits 
digits = load_digits()
```

```python
dir(digits)
```

```python
digits.target
```

```python
digits.target_names
```

```python
data = pd.DataFrame(digits.data)
data.head()
```

```python
data['target'] = digits.target
data.head()
```

```python
from sklearn.model_selection import train_test_split 
Xtrain , Xtest , ytrain , ytest = train_test_split(data.drop('target',axis='columns'),data.target , test_size=0.2 , random_state= 42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
Knear = KNeighborsClassifier(n_neighbors=5)
```

```python
len(Xtrain)
```

```python
len(Xtest)
```

```python
Knear.fit(Xtrain,ytrain)
```

```python
Knear.score(Xtest,ytest)
```

```python
from sklearn.metrics import confusion_matrix 
ypred = Knear.predict(Xtest)
confusion = confusion_matrix(ytest,ypred)
confusion
```

```python
import seaborn as sn
import matplotlib.pyplot as plt 

sn.heatmap(confusion,annot=True)
plt.ylabel('Truth')
plt.xlabel('Predicted ')
```

```python
from sklearn.metrics import classification_report 
print(classification_report(ypred,ytest))
```
